## Neste projeto, trabalharemos com dois arquivos csvs separados, contendo informações sobre o preço da gasolina no Brasil. 

### 1. Carregue os conjuntos de dados "gasolina_2000+.csv" e "gasolina_2010+.csv" em dois DataFrames diferentes e combine-os em um único DataFrame.

```python
df_petro = pd.read_csv('gasolina_2000+.csv')
df_petro2 = pd.read_csv('gasolina_2010+.csv')

df_petro = pd.concat([df_petro, df_petro2], ignore_index=True)
```

### 4. Você deve ter percebido que as colunas DATA INICIAL e DATA FINAL estão formatadas como string. Utilizando o método pd.to_datetime(), converta ambas colunas para Timestamp / Datetime.

```python
df_petro['DATA INICIAL'] = pd.to_datetime(df_petro['DATA INICIAL'], format=  '%Y-%m-%d',errors='coerce')
df_petro.info('DATA INICIAL')
```

```python
df_petro['DATA FINAL'] = pd.to_datetime(df_petro['DATA FINAL'], format=  '%Y-%m-%d',errors='coerce')
df_petro.info('DATA FINAL')
```

### 5. Crie uma nova coluna para representar o ano e o mês(aaaa-mm), utilizando a coluna DATA FINAL como referência. 

```python
df_petro['ANO-MES'] = df_petro['DATA FINAL'].dt.to_period('M')
df_petro.info('ANO-MES')
```

### 6. Utilizando o value_counts(), liste todos os tipos de produtos contidos na base de dados.

```python
df_petro['PRODUTO'].value_counts()
```

### 7. Filtre o DataFrame para obter apenas dados da 'GASOLINA COMUM'. Grave em uma nova variável.

```python
df_gasolina = df_petro[df_petro['PRODUTO'] == 'GASOLINA COMUM']
df_gasolina.head()
```

### 8. Qual o preço médio de revenda da gasolina em agosto de 2008?

```python
df_gasolina2008 = df_gasolina[df_gasolina['ANO-MES'] == '2008-08']
df_gasolina2008['PREÇO MÉDIO REVENDA'].mean()
```

### 11. Qual a média de preço dos estados da região sul em 2012?

```python
df_2012 = df_gasolina[(df_gasolina['DATA FINAL'].dt.year == 2012) & (df_gasolina['REGIÃO'] == 'SUL')]
df_2012['PREÇO MÉDIO REVENDA'].mean()
```

### 12. Você conseguiria obter uma tabela contendo a variação percentual ano a ano para o estado do Rio de Janeiro?

```python
df_rj = df_gasolina[df_gasolina['ESTADO'] == 'RIO DE JANEIRO']

precos_ano_rj = df_rj.groupby(df_rj['DATA FINAL'].dt.year)['PREÇO MÉDIO REVENDA'].mean()

variacao_ano_a_ano = precos_ano_rj.pct_change() * 100

variacao_ano_a_ano
```

### DESAFIO: Crie uma tabela contendo uma serie temporal com a diferença absoluta e percentual entre os valores mais baratos e caros. Apresente também ao lado os estados na qual os maiores e menores preços foram registrados.

```python
precos_estados_ano = df_gasolina.groupby(['ESTADO', df_gasolina['DATA FINAL'].dt.year])['PREÇO MÉDIO REVENDA'].mean()


variacao_todos = precos_estados_ano.groupby(level=0).pct_change() * 100

variacao_2021 = variacao_todos.xs(2021, level=1)

variacao_2021.sort_values(ascending=False)
```
